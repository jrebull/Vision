## 1️⃣ Introducción y Motivación

### ¿Por qué Graph Cuts?

La **segmentación de imágenes** es uno de los problemas fundamentales en visión computacional. Graph Cuts se ha convertido en uno de los métodos más poderosos y versátiles para esta tarea.

### 🎯 Aplicaciones Principales:

1. **Creación de Datasets Anotados**
   - Segmentación semi-automática para etiquetar grandes volúmenes de datos
   - Reducción drástica del tiempo de anotación manual (80-90%)
   - Alta precisión en los bordes de los objetos

2. **Edición de Imágenes**
   - Extracción de objetos (Background removal)
   - Composición de imágenes
   - Efectos visuales

3. **Medicina y Biología**
   - Segmentación de órganos en imágenes médicas
   - Análisis de células y tejidos

### 💡 ¿Por qué es tan popular para crear Datasets?

Graph Cuts es especialmente valioso en la creación de datasets porque:

- ✅ **Interactividad**: Permite correcciones rápidas con mínima interacción del usuario
- ✅ **Precisión**: Captura bordes finos y detalles complejos
- ✅ **Eficiencia**: Procesa imágenes de alta resolución en segundos
- ✅ **Consistencia**: Produce segmentaciones reproducibles
- ✅ **Flexibilidad**: Se adapta a diferentes tipos de objetos y escenas

## 2️⃣ Fundamentos Teóricos de Graph Cuts

### 📖 Artículo Base

**Interactive Graph Cuts for Optimal Boundary & Region Segmentation of Objects in N-D Images**

*Yuri Y. Boykov & Marie-Pierre Jolly*

*Siemens Corporate Research, 2001*

*IEEE ICCV 2001*

---

### 🔍 ¿En qué consiste el método?

Graph Cuts es un **método de optimización global** que formula la segmentación de imágenes como un problema de **corte mínimo en un grafo**.

#### Componentes Principales:

1. **Construcción del Grafo G = (V, E)**
   - **V (Vértices)**: Cada píxel de la imagen + 2 nodos terminales (source y sink)
   - **E (Aristas)**: Conexiones entre píxeles vecinos y conexiones a los terminales

2. **Tipos de Aristas**
   - **n-links**: Conectan píxeles vecinos (edge weights)
   - **t-links**: Conectan píxeles a source (foreground) o sink (background)

3. **Función de Energía**

   La segmentación óptima minimiza:
   
   E(A) = λ · R(A) + B(A)
   
   Donde:
   - R(A) = Término regional (penaliza asignaciones inconsistentes con las semillas)
   - B(A) = Término de borde (penaliza bordes en regiones homogéneas)
   - λ = Parámetro de balance

4. **Algoritmo de Ford-Fulkerson**
   - Encuentra el corte mínimo que separa source y sink
   - Equivalente al flujo máximo por el teorema min-cut/max-flow

### 🏆 ¿Cuándo elegir Graph Cuts?

**Elige Graph Cuts si:**
1. Necesitas **máxima precisión** en los bordes
2. Puedes proporcionar **interacción humana** (semillas)
3. El **tiempo de procesamiento** no es crítico (segundos es aceptable)
4. Trabajas en **creación de datasets** o **aplicaciones médicas**
5. Los objetos tienen **bordes relativamente claros**

**Elige otro método si:**
1. Necesitas **procesamiento automático** completo → Watershed, Clustering
2. Necesitas **velocidad extrema** → K-means, Superpixels
3. Tienes **recursos limitados** → Region Growing, K-means

## 5️⃣ Implementación del Algoritmo Graph Cuts

### 🏗️ Arquitectura del Sistema

Implementaremos Graph Cuts siguiendo principios de **Clean Code**:
- Separación de responsabilidades
- Nombres descriptivos
- Funciones pequeñas y cohesivas
- Documentación clara

```python
class GraphCutsSegmentador:
    """
    Implementación de Graph Cuts para segmentación interactiva de imágenes.

    Basado en:
    'Interactive Graph Cuts for Optimal Boundary & Region Segmentation'
    Boykov & Jolly, ICCV 2001
    """

    FOREGROUND = 1
    BACKGROUND = 0
    INFINITO = 1e10

    def __init__(self, lambda_param=50.0):
        """Inicializa el segmentador."""
        self.imagen = None
        self.semillas_fg = []
        self.semillas_bg = []
        self.mascara = None
        self.lambda_param = lambda_param

    def cargar_imagen(self, imagen):
        """Carga una imagen."""
        if isinstance(imagen, str):
            self.imagen = cv2.imread(imagen)
            if self.imagen is None:
                raise FileNotFoundError(f"No se pudo cargar: {imagen}")
            self.imagen = cv2.cvtColor(self.imagen, cv2.COLOR_BGR2RGB)
        else:
            self.imagen = imagen.copy()

        self.semillas_fg = []
        self.semillas_bg = []
        self.mascara = None

    def agregar_semilla(self, x, y, tipo, radio=3):
        """Agrega una semilla de foreground o background."""
        if self.imagen is None:
            raise ValueError("Primero debes cargar una imagen")

        h, w = self.imagen.shape[:2]

        for dy in range(-radio, radio + 1):
            for dx in range(-radio, radio + 1):
                if dx*dx + dy*dy <= radio*radio:
                    nx, ny = x + dx, y + dy
                    if 0 <= nx < w and 0 <= ny < h:
                        coord = (nx, ny)
                        if tipo == self.FOREGROUND:
                            if coord not in self.semillas_fg:
                                self.semillas_fg.append(coord)
                        else:
                            if coord not in self.semillas_bg:
                                self.semillas_bg.append(coord)

    def _calcular_peso_boundary(self, pixel1, pixel2):
        """Calcula el peso de la arista entre dos píxeles vecinos."""
        diff = np.sum((pixel1.astype(float) - pixel2.astype(float)) ** 2)
        sigma = 30.0
        peso = np.exp(-diff / (2 * sigma * sigma))
        return peso

    def segmentar(self):
        """Ejecuta el algoritmo de Graph Cuts."""
        if self.imagen is None:
            raise ValueError("Primero debes cargar una imagen")

        if not self.semillas_fg or not self.semillas_bg:
            raise ValueError("Debes agregar semillas de foreground Y background")

        print("🔨 Construyendo grafo...")
        h, w = self.imagen.shape[:2]
        num_nodos = h * w

        grafo = maxflow.Graph[float](num_nodos, num_nodos * 4)
        nodos = grafo.add_nodes(num_nodos)

        def coord_a_nodo(x, y):
            return y * w + x

        # Agregar t-links
        for y in range(h):
            for x in range(w):
                nodo_idx = coord_a_nodo(x, y)

                peso_source = 0.0
                if (x, y) in self.semillas_fg:
                    peso_source = self.INFINITO

                peso_sink = 0.0
                if (x, y) in self.semillas_bg:
                    peso_sink = self.INFINITO

                grafo.add_tedge(nodo_idx, peso_source, peso_sink)

        # Agregar n-links
        for y in range(h):
            for x in range(w):
                nodo_actual = coord_a_nodo(x, y)
                pixel_actual = self.imagen[y, x]

                if x < w - 1:
                    vecino = coord_a_nodo(x + 1, y)
                    pixel_vecino = self.imagen[y, x + 1]
                    peso = self.lambda_param * self._calcular_peso_boundary(pixel_actual, pixel_vecino)
                    grafo.add_edge(nodo_actual, vecino, peso, peso)

                if y < h - 1:
                    vecino = coord_a_nodo(x, y + 1)
                    pixel_vecino = self.imagen[y + 1, x]
                    peso = self.lambda_param * self._calcular_peso_boundary(pixel_actual, pixel_vecino)
                    grafo.add_edge(nodo_actual, vecino, peso, peso)

        print("✂️ Ejecutando max-flow/min-cut...")
        flujo = grafo.maxflow()
        print(f"📊 Flujo máximo: {flujo:.2f}")

        self.mascara = np.zeros((h, w), dtype=bool)

        for y in range(h):
            for x in range(w):
                nodo_idx = coord_a_nodo(x, y)
                self.mascara[y, x] = grafo.get_segment(nodo_idx) == 0

        print("✅ Segmentación completada")
        return self.mascara

    def visualizar_semillas(self):
        """Crea una visualización con las semillas dibujadas."""
        if self.imagen is None:
            return None

        vis = self.imagen.copy()

        for x, y in self.semillas_bg:
            cv2.circle(vis, (x, y), 2, (255, 0, 0), -1)

        for x, y in self.semillas_fg:
            cv2.circle(vis, (x, y), 2, (0, 255, 0), -1)

        return vis

print("✅ Clase GraphCutsSegmentador implementada correctamente")
```

```python
# Demo de segmentación
def demo_segmentacion_basica():
    print("🎬 Iniciando demo de segmentación...\n")

    # Cargar imagen
    segmentador = GraphCutsSegmentador(lambda_param=50.0)
    segmentador.cargar_imagen('ejemplo.jpg')
    print(f"✅ Imagen cargada: {segmentador.imagen.shape}")

    h, w = segmentador.imagen.shape[:2]

    # Agregar semillas de foreground (región central)
    print("\n📍 Agregando semillas de foreground (centro)...")
    for x in range(w//3, 2*w//3, 20):
        for y in range(h//3, 2*h//3, 20):
            segmentador.agregar_semilla(x, y, GraphCutsSegmentador.FOREGROUND, radio=3)

    # Agregar semillas de background (bordes)
    print("📍 Agregando semillas de background (bordes)...")
    for x in range(0, w, 30):
        segmentador.agregar_semilla(x, 10, GraphCutsSegmentador.BACKGROUND, radio=3)
        segmentador.agregar_semilla(x, h-10, GraphCutsSegmentador.BACKGROUND, radio=3)

    for y in range(0, h, 30):
        segmentador.agregar_semilla(10, y, GraphCutsSegmentador.BACKGROUND, radio=3)
        segmentador.agregar_semilla(w-10, y, GraphCutsSegmentador.BACKGROUND, radio=3)

    print(f"✅ Total semillas: {len(segmentador.semillas_fg)} FG, {len(segmentador.semillas_bg)} BG\n")

    # Segmentar
    mascara = segmentador.segmentar()

    # Visualizar resultados
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    axes[0, 0].imshow(segmentador.imagen)
    axes[0, 0].set_title('1. Imagen Original', fontsize=14, fontweight='bold')
    axes[0, 0].axis('off')

    axes[0, 1].imshow(segmentador.visualizar_semillas())
    axes[0, 1].set_title('2. Con Semillas (Verde=FG, Rojo=BG)', fontsize=14, fontweight='bold')
    axes[0, 1].axis('off')

    axes[1, 0].imshow(mascara, cmap='gray')
    axes[1, 0].set_title('3. Máscara de Segmentación', fontsize=14, fontweight='bold')
    axes[1, 0].axis('off')

    resultado = segmentador.imagen.copy()
    resultado[~mascara] = 0
    axes[1, 1].imshow(resultado)
    axes[1, 1].set_title('4. Resultado Final', fontsize=14, fontweight='bold')
    axes[1, 1].axis('off')

    plt.tight_layout()
    plt.show()

    pct_fg = (mascara.sum() / mascara.size) * 100
    print(f"\n📊 Resultado: {pct_fg:.1f}% foreground, {100-pct_fg:.1f}% background")
    print("✅ Segmentación completada exitosamente")

# Ejecutar demo
demo_segmentacion_basica()
```

## 7️⃣ Aplicación: Creación de Datasets

### 🏭 Pipeline de Creación de Dataset

Graph Cuts es especialmente útil para crear datasets anotados eficientemente.

```python
import json
import os

class GeneradorDataset:
    """Herramienta para crear datasets anotados usando Graph Cuts."""

    def __init__(self, directorio_salida='dataset_output'):
        self.segmentador = GraphCutsSegmentador()
        self.directorio_salida = directorio_salida
        self.anotaciones = []

        os.makedirs(f"{directorio_salida}/imagenes", exist_ok=True)
        os.makedirs(f"{directorio_salida}/mascaras", exist_ok=True)

    def procesar_imagen(self, ruta_imagen, semillas_fg, semillas_bg, nombre_salida):
        """Procesa una imagen y guarda los resultados."""
        self.segmentador.cargar_imagen(ruta_imagen)

        for x, y in semillas_fg:
            self.segmentador.agregar_semilla(x, y, GraphCutsSegmentador.FOREGROUND, radio=5)

        for x, y in semillas_bg:
            self.segmentador.agregar_semilla(x, y, GraphCutsSegmentador.BACKGROUND, radio=5)

        mascara = self.segmentador.segmentar()

        # Guardar imagen original
        cv2.imwrite(
            f"{self.directorio_salida}/imagenes/{nombre_salida}.png",
            cv2.cvtColor(self.segmentador.imagen, cv2.COLOR_RGB2BGR)
        )

        # Guardar máscara
        cv2.imwrite(
            f"{self.directorio_salida}/mascaras/{nombre_salida}_mask.png",
            (mascara * 255).astype(np.uint8)
        )

        h, w = mascara.shape
        area_fg = mascara.sum()

        metadatos = {
            'nombre': nombre_salida,
            'imagen_original': f"imagenes/{nombre_salida}.png",
            'mascara': f"mascaras/{nombre_salida}_mask.png",
            'dimensiones': {'width': w, 'height': h},
            'area_foreground_px': int(area_fg),
            'porcentaje_foreground': float(area_fg / (h * w) * 100),
            'num_semillas_fg': len(semillas_fg),
            'num_semillas_bg': len(semillas_bg),
        }

        self.anotaciones.append(metadatos)
        return metadatos

    def exportar_metadatos(self):
        """Exporta los metadatos del dataset."""
        with open(f"{self.directorio_salida}/anotaciones.json", 'w') as f:
            json.dump(self.anotaciones, f, indent=2)
        print(f"✅ Metadatos exportados")

    def generar_reporte(self):
        """Genera un reporte estadístico del dataset."""
        if not self.anotaciones:
            print("❌ No hay anotaciones")
            return

        print("\n" + "="*60)
        print("📊 REPORTE DEL DATASET")
        print("="*60)
        print(f"📁 Directorio: {self.directorio_salida}")
        print(f"📷 Total imágenes: {len(self.anotaciones)}")

        areas = [a['area_foreground_px'] for a in self.anotaciones]
        porcentajes = [a['porcentaje_foreground'] for a in self.anotaciones]

        print(f"\n📈 Estadísticas:")
        print(f"   - Promedio área FG: {np.mean(areas):.0f} px ({np.mean(porcentajes):.1f}%)")
        print(f"   - Mínimo: {np.min(areas):.0f} px ({np.min(porcentajes):.1f}%)")
        print(f"   - Máximo: {np.max(areas):.0f} px ({np.max(porcentajes):.1f}%)")
        print("="*60)

print("✅ Clase GeneradorDataset lista")
```

```python
# Demo de creación de dataset
def demo_creacion_dataset():
    print("🏭 Iniciando creación de dataset...\n")

    generador = GeneradorDataset('mi_dataset')

    img = cv2.imread('ejemplo.jpg')
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    h, w = img.shape[:2]

    semillas_fg = [(w//2 + x, h//2 + y) for x in range(-20, 20, 10) for y in range(-20, 20, 10)]
    semillas_bg = (
        [(x, 5) for x in range(0, w, 30)] +
        [(x, h-5) for x in range(0, w, 30)] +
        [(5, y) for y in range(0, h, 30)] +
        [(w-5, y) for y in range(0, h, 30)]
    )

    print("📸 Procesando imagen...")
    metadatos = generador.procesar_imagen(
        'ejemplo.jpg',
        semillas_fg,
        semillas_bg,
        'imagen_001'
    )

    print(f"   ✅ Procesada: {metadatos['porcentaje_foreground']:.1f}% foreground")

    generador.exportar_metadatos()
    generador.generar_reporte()

    # Visualizar
    img_original = cv2.imread('mi_dataset/imagenes/imagen_001.png')
    img_original = cv2.cvtColor(img_original, cv2.COLOR_BGR2RGB)

    mascara = cv2.imread('mi_dataset/mascaras/imagen_001_mask.png', cv2.IMREAD_GRAYSCALE)

    fig, axes = plt.subplots(1, 3, figsize=(15, 5))

    axes[0].imshow(img_original)
    axes[0].set_title('Original', fontsize=14, fontweight='bold')
    axes[0].axis('off')

    axes[1].imshow(mascara, cmap='gray')
    axes[1].set_title('Máscara (Ground Truth)', fontsize=14, fontweight='bold')
    axes[1].axis('off')

    resultado = img_original.copy()
    resultado[mascara == 0] = 0
    axes[2].imshow(resultado)
    axes[2].set_title('Segmentado', fontsize=14, fontweight='bold')
    axes[2].axis('off')

    plt.tight_layout()
    plt.show()

    print("\n🎉 Dataset creado exitosamente!")

demo_creacion_dataset()
```

## 8️⃣ Casos de Uso y Limitaciones

### ✅ Casos de Éxito: Cuándo usar Graph Cuts

1. **Segmentación Médica** - Delimitar tumores, órganos en MRI/CT
2. **Edición Profesional** - Background removal, composición
3. **Creación de Datasets** - Anotar miles de imágenes eficientemente

1. **Objetos con Bordes Difusos** - Humo, niebla, llamas
2. **Bajo Contraste Global** - Objeto similar al fondo
3. **Tiempo Real** - Video streaming, AR en vivo
4. **Completamente Automático** - Sin posibilidad de interacción
5. **Múltiples Clases** - Segmentación semántica compleja

## 9️⃣ Análisis interactivo con Graph Cuts


### Segmentación con GrabCut y su relación con GraphCut 🧩✂️

GrabCut es un método de segmentación basado en GraphCut. GraphCut representa la imagen como un grafo donde cada píxel es un nodo, y la separación entre objeto y fondo se obtiene encontrando el corte mínimo. GrabCut agrega un proceso iterativo: el usuario marca una región aproximada y el algoritmo refina automáticamente la segmentación. Es útil para aislar objetos de forma semiautomática. 🎯


A continuación, se implementará el uso de GrabCut para segmentar imágenes.

### Funciones auxiliares

📁 `download_drive_folder_public`: Descarga una carpeta pública de Google Drive usando su URL o su ID. Crea la carpeta de destino si no existe y guarda ahí todos los archivos.

📂 `list_folder_contents`: Recorre una carpeta de forma recursiva y devuelve una lista ordenada con las rutas de todos los archivos que contiene.

🖼️ `redimensionar_y_guardar`: Abre cada imagen, detecta si es horizontal o vertical, la redimensiona según corresponda y la guarda en la carpeta de salida con nombres consecutivos.

📸 `mostrar_imagenes_horizontal`: Muestra varias imágenes en una misma fila. Cada una aparece sin ejes y con un título que indica su índice.


```python
def download_drive_folder_public(folder_url_or_id: str, dest_dir: str):
    os.makedirs(dest_dir, exist_ok=True)
    if re.search(r'/folders/([a-zA-Z0-9_-]{10,})', folder_url_or_id):
        url = folder_url_or_id
    else:
        url = f"https://drive.google.com/drive/folders/{folder_url_or_id}"
    gdown.download_folder(url=url, output=dest_dir, quiet=False, use_cookies=False, remaining_ok=True)
    print(f"Done. Files saved under: {os.path.abspath(dest_dir)}")

def list_folder_contents(base_folder: str) -> list:
    file_list = []
    for root, _, files in os.walk(base_folder):
        for f in files:
            relative_path = os.path.join(root, f)
            file_list.append(relative_path)
    return sorted(file_list)

def redimensionar_y_guardar(rutas, carpeta_salida,
                            out_horizontal=(640,480),
                            out_vertical=(480,640)):
    os.makedirs(carpeta_salida, exist_ok=True)
    for i, ruta in enumerate(rutas):
        img = Image.open(ruta)
        w, h = img.size
        if w > h:
            new_size = out_horizontal
        else:
            new_size = out_vertical

        img_res = img.resize(new_size, Image.LANCZOS)
        nombre = f"img_{i}.jpg"
        destino = os.path.join(carpeta_salida, nombre)
        img_res.save(destino, "JPEG")
        print(f"Guardada: {destino}")

def mostrar_imagenes_horizontal(rutas):
    n = len(rutas)
    fig, axes = plt.subplots(1, n, figsize=(4*n, 5))
    if n == 1:
        axes = [axes]
    for i, (ax, ruta) in enumerate(zip(axes, rutas)):
        img = np.array(Image.open(ruta))
        ax.imshow(img)
        ax.set_title(f"[{i}]")
        ax.axis("off")
    plt.tight_layout()
    plt.show()
```

### Etapa 1: Definición de las funciones de GrabCut

A continuación se describen las dos funciones que aplican GrabCut usando un rectángulo o un círculo como guía inicial.

---

### `graphcut_rectangulo(...)` 🟥

- Carga la imagen elegida.  
- Construye un rectángulo a partir de las coordenadas.  
- Inicializa la máscara y los modelos requeridos por GrabCut.  
- Ejecuta GrabCut usando el rectángulo como referencia del objeto.  
- Produce la máscara final y la aplica a la imagen.  
- Muestra: original, original con rectángulo, máscara y resultado.  

---

### `graphcut_circulo(...)` 🔵

- Carga la imagen elegida.  
- Construye una máscara circular marcando la región probable del objeto.  
- Inicializa la máscara y los modelos como en la versión rectangular.  
- Ejecuta GrabCut usando la máscara circular en vez de un rectángulo.  
- Genera la máscara final y el recorte segmentado.  
- Muestra: original, original con círculo, máscara y resultado.  

---

### Diferencias principales ⚖️

- Una usa un rectángulo explícito. La otra usa una máscara circular.  
- La versión rectangular usa inicialización con rectángulo. La circular usa evaluación sobre la máscara.  
- Cambia la guía visual: rectángulo con coordenadas vs. círculo con centro y radio.  
- La rectangular es más práctica cuando el objeto encaja en una caja. La circular es más intuitiva cuando la región es radial o redondeada.  


```python
def graphcut_rectangulo(numero_imagen, x0, y0, x1, y1, iteraciones=5):

    # Carga la imagen desde la ruta seleccionada
    IMG_PATH = rutas_imagenes[numero_imagen]
    IMG = iio.imread(IMG_PATH)

    # Calcula el ancho y alto del rectángulo
    w = x1 - x0
    h = y1 - y0

    # Muestra el tamaño máximo permitido de coordenadas
    print("Máximo X:", IMG.shape[1])
    print("Máximo Y:", IMG.shape[0])

    # Prepara la máscara y los modelos de fondo y frente para GrabCut
    rows, cols, _ = IMG.shape
    mask = np.zeros((rows, cols), np.uint8)            # aquí GrabCut guardará etiquetas
    bg_model = np.zeros((1, 65), np.float64)           # modelo estadístico del fondo
    fg_model = np.zeros((1, 65), np.float64)           # modelo estadístico del frente
    rect = (x0, y0, w, h)                              # rectángulo inicial

    # Ejecuta GrabCut usando el rectángulo como guía
    cv2.grabCut(IMG, mask, rect, bg_model, fg_model,
               iteraciones, cv2.GC_INIT_WITH_RECT)

    # Convierte la máscara a valores binarios (1 objeto, 0 fondo)
    mask_graph_cut = np.where((mask == 2) | (mask == 0), 0, 1).astype('uint8')

    # Aplica la máscara sobre la imagen para obtener el resultado final
    img_result = IMG * mask_graph_cut[:, :, np.newaxis]

    # Prepara una figura con 4 vistas en horizontal
    fig, axes = plt.subplots(1, 4, figsize=(20, 5))

    # 1. Imagen original
    axes[0].imshow(IMG)
    axes[0].set_title("Original")
    axes[0].axis("off")

    # 2. Imagen con el rectángulo dibujado
    axes[1].imshow(IMG)
    rect_patch = patches.Rectangle((x0, y0), w, h,
                                   linewidth=2, edgecolor='red', facecolor='none')
    axes[1].add_patch(rect_patch)

    # Coordenadas visibles para referencia
    axes[1].text(x0-100, y0-5, f"({x0},{y0})", color="red", fontsize=12)
    axes[1].text(x1-50, y1+40, f"({x1},{y1})", color="red", fontsize=12)

    axes[1].set_title("Con rectángulo")
    axes[1].grid(True)

    # 3. Máscara generada por GrabCut
    axes[2].imshow(mask)
    axes[2].set_title("Mask")
    axes[2].axis("off")

    # 4. Resultado final segmentado
    axes[3].imshow(img_result)
    axes[3].set_title("Resultado")
    axes[3].axis("off")

    plt.tight_layout()
    plt.show()

    return None
```

```python
def graphcut_circulo(numero_imagen, cx, cy, r, iteraciones=5):
    """
    Esta variante usa GrabCut partiendo de una máscara circular.
    El círculo indica la zona donde probablemente está el objeto.
    GrabCut ajusta esa región mediante su modelo de GraphCut.
    """

    # Carga la imagen seleccionada
    IMG_PATH = rutas_imagenes[numero_imagen]
    IMG = iio.imread(IMG_PATH)

    # Crea la máscara y los modelos internos para GrabCut
    rows, cols, _ = IMG.shape
    mask = np.zeros((rows, cols), np.uint8)            # aquí se guardan etiquetas de fondo/objeto
    bg_model = np.zeros((1, 65), np.float64)           # modelo estadístico del fondo
    fg_model = np.zeros((1, 65), np.float64)           # modelo estadístico del frente

    # Construye una máscara circular
    # Cada punto dentro del radio se marca como "probable foreground"
    Y, X = np.ogrid[:rows, :cols]
    dist = (X - cx)**2 + (Y - cy)**2
    mask[dist <= r*r] = cv2.GC_PR_FGD

    # Ejecuta GrabCut usando la máscara como guía inicial
    cv2.grabCut(IMG, mask, None, bg_model, fg_model,
               iteraciones, cv2.GC_EVAL)

    # Convierte la máscara a binaria (1 objeto, 0 fondo)
    mask_graph_cut = np.where((mask == 2) | (mask == 0), 0, 1).astype('uint8')

    # Obtiene la imagen segmentada aplicando la máscara
    img_result = IMG * mask_graph_cut[:, :, np.newaxis]

    # Visualiza todas las etapas en 4 columnas
    fig, axes = plt.subplots(1, 4, figsize=(20, 5))

    # 1. Imagen original
    axes[0].imshow(IMG)
    axes[0].set_title("Original")
    axes[0].axis("off")

    # 2. Imagen con el círculo usado como guía
    axes[1].imshow(IMG)
    circle = plt.Circle((cx, cy), r, color='red', fill=False, linewidth=2)
    axes[1].add_patch(circle)
    axes[1].grid(True)
    axes[1].set_title("Con círculo")

    # 3. Máscara producida por GrabCut
    axes[2].imshow(mask)
    axes[2].set_title("Mask")
    axes[2].axis("off")

    # 4. Resultado segmentado
    axes[3].imshow(img_result)
    axes[3].set_title("Resultado")
    axes[3].axis("off")

    plt.tight_layout()
    plt.show()

    return None
```

### ¿Y qué pasa si combinamos ambas? 🟥🔵✨

La función `graphcut_custom` lleva la idea del playground un paso más allá. En vez de elegir solo rectángulos o solo círculos, permite combinar múltiples figuras a la vez y construir una máscara “a medida” para guiar a GrabCut. Esto da mucha más flexibilidad para marcar regiones complejas del objeto. 🎯

---

 `graphcut_custom(numero_imagen, rects=None, circles=None, iteraciones=5)`

- 📥 **Entrada principal**
  - `numero_imagen`: índice para elegir la imagen desde `rutas_imagenes`.
  - `rects`: lista de rectángulos `[(x0, y0, x1, y1), ...]`.  
  - `circles`: lista de círculos `[(cx, cy, r), ...]`.  
  - `iteraciones`: número de iteraciones de GrabCut.

- 🏞 **Carga e inicialización**
  - Carga la imagen `IMG` a partir de `numero_imagen`.
  - Obtiene dimensiones `rows`, `cols`.
  - Crea una `mask` llena de fondo (valor por defecto: todo se considera background).
  - Inicializa `bg_model` y `fg_model`, modelos de fondo y primer plano que usa GrabCut.

- 🟥 **Marcado de rectángulos como probable objeto**
  - Recorre cada `(x0, y0, x1, y1)` en `rects`.
  - Clampa las coordenadas para que no se salgan de la imagen.
  - Marca en la máscara esas zonas como `cv2.GC_PR_FGD` (probable foreground).
  - Esto le dice a GrabCut: “en estas áreas es muy probable que haya objeto”.

- 🔵 **Marcado de círculos como probable objeto**
  - Calcula una malla de coordenadas `X`, `Y` con `np.ogrid`.
  - Para cada círculo `(cx, cy, r)` calcula la distancia al centro.
  - Donde `dist <= r*r`, marca la máscara también como `cv2.GC_PR_FGD`.
  - Así, rectángulos y círculos se combinan sobre la misma máscara.

- ⚙️ **Ejecución de GrabCut con máscara personalizada**
  - Llama a `cv2.grabCut` pasando:
    - La imagen `IMG`.
    - La máscara ya rellenada con las áreas marcadas.
    - `None` para el rectángulo porque se usa la máscara.
    - `bg_model`, `fg_model` y el número de iteraciones.
    - El modo `cv2.GC_INIT_WITH_MASK`, indicando que la guía viene de la máscara, no de un rectángulo.

- 🧪 **Construcción de la máscara binaria final**
  - Convierte la máscara de estados de GrabCut a binario:
    - 1 para `GC_FGD` y `GC_PR_FGD` (objeto y probable objeto).
    - 0 para el resto (fondo).
  - Genera `img_result` multiplicando la imagen original por esta máscara 3D.

---

### ¿Por qué es útil esta combinación? 🚀

- Permite marcar diferentes partes del objeto con distintas figuras en una misma llamada.  
- Es más flexible que usar un único rectángulo o un único círculo.  
- Facilita la segmentación de formas complejas (por ejemplo, un objeto alargado más una parte redondeada).  
- Mantiene la lógica de GrabCut, pero da mucho más control al usuario sobre la máscara inicial.  


```python
def graphcut_custom(numero_imagen, rects=None, circles=None, iteraciones=5):
    """
    Segmenta usando GrabCut a partir de una máscara 'custom' formada
    por varios rectángulos y varios círculos.

    rects  = lista de rectángulos [(x0, y0, x1, y1), ...]
    circles = lista de círculos [(cx, cy, r), ...]
    """

    if rects is None:
        rects = []
    if circles is None:
        circles = []

    # Carga imagen
    IMG_PATH = rutas_imagenes[numero_imagen]
    IMG = iio.imread(IMG_PATH)

    rows, cols, _ = IMG.shape
    mask = np.zeros((rows, cols), np.uint8)      # todo fondo por defecto
    bg_model = np.zeros((1, 65), np.float64)
    fg_model = np.zeros((1, 65), np.float64)

    # Marca rectángulos como "probable foreground"
    for (x0, y0, x1, y1) in rects:
        x0c, x1c = max(0, x0), min(cols, x1)
        y0c, y1c = max(0, y0), min(rows, y1)
        mask[y0c:y1c, x0c:x1c] = cv2.GC_PR_FGD

    # Marca círculos como "probable foreground"
    Y, X = np.ogrid[:rows, :cols]
    for (cx, cy, r) in circles:
        dist = (X - cx)**2 + (Y - cy)**2
        mask[dist <= r*r] = cv2.GC_PR_FGD

    # Ejecuta GrabCut usando la máscara como guía
    cv2.grabCut(
        IMG,
        mask,
        None,
        bg_model,
        fg_model,
        iteraciones,
        cv2.GC_INIT_WITH_MASK  # aquí usamos la máscara, no un rectángulo
    )

    # Binaria: 1 objeto, 0 fondo
    mask_graph_cut = np.where(
        (mask == cv2.GC_FGD) | (mask == cv2.GC_PR_FGD),
        1,
        0
    ).astype("uint8")

    img_result = IMG * mask_graph_cut[:, :, np.newaxis]

    # Visualización
    fig, axes = plt.subplots(1, 4, figsize=(20, 5))

    # 1. Original
    axes[0].imshow(IMG)
    axes[0].set_title("Original")
    axes[0].axis("off")

    # 2. Imagen con todas las figuras dibujadas
    axes[1].imshow(IMG)
    for (x0, y0, x1, y1) in rects:
        w = x1 - x0
        h = y1 - y0
        rect_patch = patches.Rectangle(
            (x0, y0), w, h,
            linewidth=2, edgecolor='red', facecolor='none'
        )
        axes[1].add_patch(rect_patch)
    for (cx, cy, r) in circles:
        circle = plt.Circle((cx, cy), r, color='lime', fill=False, linewidth=2)
        axes[1].add_patch(circle)
    axes[1].set_title("Figuras guía")
    axes[1].grid(True)

    # 3. Máscara de GrabCut
    axes[2].imshow(mask)
    axes[2].set_title("Mask")
    axes[2].axis("off")

    # 4. Resultado final
    axes[3].imshow(img_result)
    axes[3].set_title("Resultado")
    axes[3].axis("off")

    plt.tight_layout()
    plt.show()

    return None
```

```python
rectangulos = [
    (100, 200, 350, 380),
    (180,525,350,640)
]

circulos = [
    (220, 450, 120),
    (350, 450, 100),
    (450, 420, 100),

]

graphcut_custom(
    numero_imagen=1,
    rects=rectangulos,
    circles=circulos,
    iteraciones=5
)
```

### Resumen de la implementación 🧩

Se construyó un pequeño entorno para experimentar con segmentación basada en **GrabCut**, un algoritmo que deriva del método **GraphCut**. La idea central es modelar la imagen como un grafo donde cada píxel es un nodo y el algoritmo decide el corte óptimo que separa objeto y fondo. GrabCut mejora esta idea combinando ese corte con modelos estadísticos de color y un proceso iterativo que refina progresivamente la máscara.

Se implementaron tres variantes:

1. **graphcut_rectangulo**  
   Usa un rectángulo como aproximación inicial del objeto. GrabCut interpreta esa área como probable primer plano. Es útil cuando el objeto encaja razonablemente dentro de una caja.

2. **graphcut_circulo**  
   Genera una máscara circular que marca una zona probable de primer plano. Se diferencia porque no usa un rectángulo, sino una máscara inicial, y se apoya más en la geometría radial. Es adecuada para objetos centrados o regiones curvas.

3. **graphcut_custom**  
   Combina múltiples rectángulos y círculos al mismo tiempo. La máscara final se construye mezclando todas las figuras marcadas como probable foreground. Usa el modo de inicialización con máscara para ejecutar GrabCut. Esta variante es la más flexible y facilita capturar objetos de formas más complejas.

Las principales diferencias entre las funciones se centran en la **forma de inicialización** del algoritmo: rectángulo, círculo o máscara compuesta. Esto afecta el comportamiento de GrabCut porque el refinamiento depende mucho de la calidad de la guía inicial. También cambia la forma en que se visualiza y se interpreta la segmentación, permitiendo comparar cómo varía el resultado según las figuras utilizadas.

---

### Reflexión 🧠📌

- GrabCut es sensible a la máscara inicial. Si la región marcada representa bien la zona del objeto, el resultado mejora notablemente.  
- Las tres funciones permiten entender cómo la geometría de la selección afecta la convergencia del algoritmo.  
- La versión personalizada es la más potente porque permite guiar a GrabCut con una máscara compuesta, dando mejor control en objetos irregulares.  
- Explorar las iteraciones y las figuras guía ayuda a comprender mejor cómo trabaja la segmentación basada en energía y cortes mínimos.  
- Este mini playground sirve como base sólida para crear herramientas más avanzadas de segmentación asistida y para estudiar más a fondo el comportamiento de GraphCut y GrabCut.

## Aprendizajes del Proyecto

Graph Cuts transforma el problema de segmentación en uno de optimización en grafos, ofreciendo garantías de optimalidad global que la mayoría de métodos alternativos no proporcionan. A lo largo de este proyecto, comprobamos empíricamente que su capacidad para producir bordes precisos con mínima interacción del usuario lo hace invaluable en aplicaciones donde la calidad es crítica.

Nuestra experiencia práctica con 9 tipos distintos de imágenes —desde casos triviales como el perfil de Trump con fondo uniforme hasta casos extremadamente complejos como el pez payaso en anémona o la mamografía médica— reveló que **no existe una configuración universal de parámetros**. El éxito de la segmentación depende críticamente de:

* **Comprensión del contenido**: Identificar si los colores objeto/fondo son similares (Beta bajo) o contrastantes (Beta alto).
* **Análisis de textura**: Fondos complejos requieren K alto para ignorar ruido; fondos uniformes permiten K bajo para mayor precisión.
* **Marcado estratégico de semillas**: En casos difíciles (Simba, mamografía, pez), la cantidad y distribución de semillas es tan importante como los parámetros algorítmicos.
* **Trade-offs calidad-velocidad**: GMM + GrabCut ofrecen ~20% mejor calidad a costa de 3x tiempo de procesamiento.

El desarrollo de la interfaz parametrizable demostró ser crucial. La capacidad de ajustar K (1-200), Beta (1-50), componentes GMM (2-5), y el radio del Guided Filter en tiempo real permitió **experimentación iterativa** que aceleró nuestro entendimiento del algoritmo exponencialmente comparado con implementaciones de caja negra.

## Retos Técnicos Superados

Los desafíos más significativos no provinieron del algoritmo Graph Cuts en sí, sino de su integración con tecnologías complementarias:

1.  **Dependencias inconsistentes**: El módulo `opencv-contrib-python` para Guided Filter no está disponible universalmente, requiriendo fallbacks robustos.
2.  **Gestión de memoria**: Imágenes de alta resolución (>3000px) saturaban la memoria durante construcción del grafo, motivando el auto-downsampling a 900px.
3.  **Balance precision-performance**: GrabCut mejora bordes pero añade latencia considerable; implementamos como opción configurable.
4.  **Casos límite**: Imágenes en escala de grises (mamografía) pierden la ventaja del espacio de color Lab, exponiendo limitaciones inherentes del método.

La implementación de la función de generación de stickers con PIL demostró la versatilidad del pipeline: la misma máscara de segmentación sirve tanto para análisis científico como para aplicaciones creativas.

Graph Cuts tiene una fundamentación impecable en teoría de grafos y optimización combinatoria, pero su aplicación efectiva requiere intuición sobre el dominio del problema, experimentación sistemática con parámetros, y comprensión profunda de las limitaciones inherentes.

La experiencia de enfrentarnos a fallos iniciales —como el pez donde solo se capturaban las semillas, o Simba donde se incluía todo el piso de madera— fue invaluable. Cada fallo nos forzó a revisar nuestras suposiciones, entender mejor los términos de energía, y apreciar por qué ciertos parámetros funcionan en ciertos contextos.

El trabajo colaborativo fue esencial: debuggear código, diseñar la UI, experimentar con parámetros y documentar resultados requirieron perspectivas complementarias. **Ningún miembro del equipo podría haber completado este proyecto individualmente con la misma calidad**.

Finalmente, el proyecto reforzó una verdad fundamental en visión computacional: **no existen algoritmos perfectos universales**. Graph Cuts es extraordinario para ciertos problemas y mediocre para otros. Saber cuándo aplicarlo y cuándo buscar alternativas es tan importante como dominar su implementación.
