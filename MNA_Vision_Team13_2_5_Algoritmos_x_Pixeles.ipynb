
````python
def display_exif_tabulate(image_path):
    img_pil = Image.open(image_path)
    exif_data = img_pil._getexif()

    if not exif_data:
        print("No se encontraron datos EXIF")
        return

    # Preparar datos para la tabla
    table_data = []
    for tag_id, value in exif_data.items():
        tag = TAGS.get(tag_id, tag_id)

        # Limitar longitud de valores muy largos
        value_str = str(value)
        if len(value_str) > 50:
            value_str = textwrap.fill(value_str, width=50)

        # Agregar emojis para ciertas categorías
        emoji = ""
        if 'Date' in tag:
            emoji = "📅"
        elif 'GPS' in tag:
            emoji = "📍"
        elif tag in ['Make', 'Model']:
            emoji = "📷"
        elif tag in ['ISO', 'ExposureTime', 'FNumber']:
            emoji = "⚙️"

        table_data.append([f"{emoji} {tag}", value_str])

    # Mostrar tabla
    print("\n" + "="*60)
    print("📸 DATOS EXIF DE LA IMAGEN 📸".center(60))
    print("="*60 + "\n")

    print(tabulate(table_data,
                   headers=['Propiedad', 'Valor'],
                   tablefmt='fancy_grid',
                   colalign=('left', 'left')))

# Usar:
display_exif_tabulate('2017_Brockton_Massachusetts')
````

## Ecualización de Histograma<a class="anchor" id="hist" ></a>

Para evaluar el contraste y la riqueza informativa de una imagen, es útil examinar la distribución de sus valores de píxeles. Una distribución relativamente uniforme sugiere buen contraste, lo que facilita la identificación de patrones, bordes y detalles relevantes para el análisis visual.
Una forma sencilla y efectiva de verificar esta distribución es mediante la generación de un histograma. Este gráfico muestra la frecuencia de cada nivel de intensidad en la imagen, permitiendo detectar rápidamente zonas sobreexpuestas, subexpuestas o con pérdida de detalle.


````python
plt.figure(figsize=(10, 6))
plt.hist(original_image.ravel(), bins=256, color='slateblue', alpha=0.7, edgecolor='black')

media_intensidad = np.mean(original_image)

plt.title("Histograma", fontsize=14)
plt.xlabel("Valor de Intensidad (0–255)", fontsize=12)
plt.ylabel("Frecuencia", fontsize=12)
ax = plt.gca()
ax.text(0.98, 0.98, f'μ = {media_intensidad:.1f}', transform=ax.transAxes,
        fontsize=12, color='navy', ha='right', va='top')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
````

Otra forma efectiva de visualizar las propiedades estadísticas de una imagen es a través de un gráfico de función de distribución acumulativa (CDF), que muestra la frecuencia acumulativa de las intensidades de los píxeles a lo largo del rango completo de 0 a 255. Esta representación ayuda a revelar cómo se distribuyen los valores de brillo y si la imagen favorece tonos más oscuros, más claros o tonos equilibrados.

````python
plt.figure(figsize=(10, 6))
plt.hist(original_image.ravel(), bins=256, color='slateblue', alpha=0.7, edgecolor='black',cumulative=True)

media_intensidad = np.mean(original_image)

plt.title("Función de distribución acumulativa (CDF)", fontsize=14)
plt.xlabel("Valor de Intensidad (0–255)", fontsize=12)
plt.ylabel("Frecuencia", fontsize=12)
ax = plt.gca()
ax.text(0.98, 0.98, f'μ = {media_intensidad:.1f}', transform=ax.transAxes,
        fontsize=12, color='navy', ha='right', va='top')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
````

````python
def plot_minimalist_histogram(image_path):
    # Configuración minimalista
    plt.rcParams['figure.facecolor'] = '#f8f9fa'
    plt.rcParams['axes.facecolor'] = '#ffffff'
    plt.rcParams['font.family'] = 'sans-serif'
    plt.rcParams['font.sans-serif'] = ['Arial', 'DejaVu Sans']

    # Cargar imagen
    img = Image.open(image_path)
    img_array = np.array(img)

    # Crear figura
    fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    fig.patch.set_facecolor('#f8f9fa')

    # Configuración de colores suaves
    colors = {
        'R': {'main': '#ff6b6b', 'light': '#ffe0e0'},
        'G': {'main': '#51cf66', 'light': '#d0f4d0'},
        'B': {'main': '#339af0', 'light': '#d0e7ff'}
    }

    titles = ['Canal Rojo', 'Canal Verde', 'Canal Azul']

    for idx, (ax, channel_name, title) in enumerate(zip(axes, ['R', 'G', 'B'], titles)):
        # Obtener datos del canal
        channel_data = img_array[:,:,idx].ravel()

        # Crear histograma con estilo elegante
        n, bins, patches = ax.hist(channel_data, bins=256,
                                   color=colors[channel_name]['light'],
                                   edgecolor=colors[channel_name]['main'],
                                   linewidth=0.5, alpha=0.7)

        # Agregar línea suave de tendencia
        from scipy.ndimage import gaussian_filter1d
        smoothed = gaussian_filter1d(n, sigma=2)
        bin_centers = (bins[:-1] + bins[1:]) / 2
        ax.plot(bin_centers, smoothed, color=colors[channel_name]['main'],
               linewidth=2, alpha=0.9)

        # Estilizar ejes
        ax.set_title(title, fontsize=12, fontweight='600', pad=15)
        ax.set_xlabel('Valor de Intensidad (0–255)', fontsize=10)
        ax.set_ylabel('Frecuencia', fontsize=10)

        # Remover bordes superiores y derechos
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
        ax.spines['left'].set_linewidth(0.5)
        ax.spines['bottom'].set_linewidth(0.5)

        # Grid sutil
        ax.grid(True, alpha=0.1, linestyle='-', linewidth=0.5)
        ax.set_axisbelow(True)

        # Agregar estadística clave de forma sutil
        mean_val = np.mean(channel_data)
        ax.text(0.98, 0.98, f'μ = {mean_val:.1f}',
               transform=ax.transAxes,
               fontsize=10, ha='right', va='top',
               bbox=dict(boxstyle='round,pad=0.3',
                        facecolor='white',
                        edgecolor=colors[channel_name]['main'],
                        alpha=0.8))

    # Título general
    fig.suptitle(f'Análisis de Histograma: {image_path}',
                fontsize=14, fontweight='bold', y=1.05)

    plt.tight_layout()
    plt.show()

# Usar
plot_minimalist_histogram('2017_Brockton_Massachusetts')
````

````python

def plot_cumulative_histogram_beautiful(image_path):
    # Configuración minimalista
    plt.rcParams['figure.facecolor'] = '#f8f9fa'
    plt.rcParams['axes.facecolor'] = '#ffffff'
    plt.rcParams['font.family'] = 'sans-serif'
    plt.rcParams['font.sans-serif'] = ['Arial', 'DejaVu Sans']

    # Cargar imagen
    img = Image.open(image_path)
    img_array = np.array(img)

    # Crear figura con subplots
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    fig.patch.set_facecolor('#f8f9fa')

    # Configuración de colores suaves
    colors = {
        'R': {'main': '#ff6b6b', 'light': '#ffe0e0', 'dark': '#e55555'},
        'G': {'main': '#51cf66', 'light': '#d0f4d0', 'dark': '#40b855'},
        'B': {'main': '#339af0', 'light': '#d0e7ff', 'dark': '#228ae6'},
        'All': {'main': '#868e96', 'light': '#e9ecef', 'dark': '#495057'}
    }

    # ============= SUBPLOT 1: Histograma Acumulativo Combinado =============
    ax1 = axes[0, 0]

    # Datos para todos los canales
    for idx, (channel_name, title) in enumerate(zip(['R', 'G', 'B'],
                                                     ['Rojo', 'Verde', 'Azul'])):
        channel_data = img_array[:,:,idx].ravel()

        # Crear histograma acumulativo
        counts, bins, patches = ax1.hist(channel_data, bins=256,
                                         cumulative=True,
                                         density=True,
                                         histtype='step',
                                         color=colors[channel_name]['main'],
                                         linewidth=2.5,
                                         alpha=0.8,
                                         label=f'Canal {title}')

    ax1.set_title('Distribución Acumulativa RGB', fontsize=13, fontweight='600', pad=15)
    ax1.set_xlabel('Intensidad de Píxel', fontsize=11)
    ax1.set_ylabel('Probabilidad Acumulada', fontsize=11)
    ax1.legend(loc='lower right', frameon=True, fancybox=True, shadow=True)
    ax1.grid(True, alpha=0.15, linestyle='-', linewidth=0.5)
    ax1.set_xlim([0, 255])
    ax1.set_ylim([0, 1.05])

    # Agregar líneas de referencia
    ax1.axhline(y=0.5, color='gray', linestyle='--', alpha=0.3, linewidth=1)
    ax1.axhline(y=0.25, color='gray', linestyle=':', alpha=0.2, linewidth=1)
    ax1.axhline(y=0.75, color='gray', linestyle=':', alpha=0.2, linewidth=1)

    # ============= SUBPLOT 2: Histograma Acumulativo Total (Luminosidad) =============
    ax2 = axes[0, 1]

    # Calcular luminosidad
    luminosity = np.mean(img_array, axis=2).ravel()

    # Histograma acumulativo con relleno
    counts, bins, patches = ax2.hist(luminosity, bins=256,
                                     cumulative=True,
                                     density=True,
                                     color=colors['All']['light'],
                                     edgecolor=colors['All']['dark'],
                                     linewidth=1.5,
                                     alpha=0.7)

    # Línea suavizada
    bin_centers = (bins[:-1] + bins[1:]) / 2
    smoothed = gaussian_filter1d(counts, sigma=1)
    ax2.plot(bin_centers, smoothed, color=colors['All']['dark'],
            linewidth=3, alpha=0.9, label='Curva suavizada')

    # Encontrar percentiles
    percentiles = [25, 50, 75]
    for p in percentiles:
        val = np.percentile(luminosity, p)
        ax2.axvline(x=val, color='red', linestyle='--', alpha=0.5, linewidth=1)
        ax2.text(val, 0.05, f'P{p}', rotation=90, fontsize=9,
                ha='right', va='bottom', color='red', alpha=0.7)

    ax2.set_title('Distribución Acumulativa de Luminosidad', fontsize=13, fontweight='600', pad=15)
    ax2.set_xlabel('Intensidad de Píxel', fontsize=11)
    ax2.set_ylabel('Probabilidad Acumulada', fontsize=11)
    ax2.legend(loc='lower right', frameon=True, fancybox=True, shadow=True)
    ax2.grid(True, alpha=0.15, linestyle='-', linewidth=0.5)
    ax2.set_xlim([0, 255])
    ax2.set_ylim([0, 1.05])

    # ============= SUBPLOTS 3-5: Canales Individuales con Área =============
    channels_data = [
        (img_array[:,:,0].ravel(), 'R', 'Canal Rojo'),
        (img_array[:,:,1].ravel(), 'G', 'Canal Verde'),
        (img_array[:,:,2].ravel(), 'B', 'Canal Azul')
    ]

    for idx, (data, color_key, title) in enumerate(channels_data):
        if idx < 2:
            ax = axes[1, idx]
        else:
            # Usar el último subplot disponible para el canal azul
            ax = plt.subplot2grid((2, 4), (1, 2), colspan=1, fig=fig)

        # Calcular histograma acumulativo
        counts, bins = np.histogram(data, bins=256)
        cumulative = np.cumsum(counts)
        cumulative = cumulative / cumulative[-1]  # Normalizar

        # Plot con relleno gradiente
        bin_centers = (bins[:-1] + bins[1:]) / 2

        ax.fill_between(bin_centers, 0, cumulative,
                       color=colors[color_key]['light'],
                       alpha=0.5)
        ax.plot(bin_centers, cumulative,
               color=colors[color_key]['main'],
               linewidth=2.5, alpha=0.9)

        # Agregar puntos en cuartiles
        quartiles = [25, 50, 75]
        for q in quartiles:
            val = np.percentile(data, q)
            y_val = np.interp(val, bin_centers, cumulative)
            ax.scatter(val, y_val, color=colors[color_key]['dark'],
                      s=50, zorder=5, alpha=0.8)
            ax.annotate(f'{q}%', xy=(val, y_val),
                       xytext=(5, 5), textcoords='offset points',
                       fontsize=9, color=colors[color_key]['dark'],
                       fontweight='bold')

        # Estadísticas
        mean_val = np.mean(data)
        median_val = np.median(data)
        std_val = np.std(data)

        stats_text = f'μ={mean_val:.1f}\nσ={std_val:.1f}'
        ax.text(0.98, 0.05, stats_text,
               transform=ax.transAxes,
               fontsize=10, ha='right', va='bottom',
               bbox=dict(boxstyle='round,pad=0.5',
                        facecolor='white',
                        edgecolor=colors[color_key]['main'],
                        alpha=0.9))

        ax.set_title(title, fontsize=12, fontweight='600', pad=10)
        ax.set_xlabel('Intensidad', fontsize=10)
        ax.set_ylabel('Probabilidad Acumulada', fontsize=10)
        ax.set_xlim([0, 255])
        ax.set_ylim([0, 1.05])

        # Estilo minimalista
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
        ax.spines['left'].set_linewidth(0.5)
        ax.spines['bottom'].set_linewidth(0.5)
        ax.grid(True, alpha=0.1, linestyle='-', linewidth=0.5)
        ax.set_axisbelow(True)

    # Título general
    fig.suptitle(f'📊 Análisis de Distribución Acumulativa: {image_path}',
                fontsize=15, fontweight='bold', y=1.02)

    plt.tight_layout()
    plt.show()

# Usar
plot_cumulative_histogram_beautiful('2017_Brockton_Massachusetts')
````

> ### 📌 **Definición**
> **El método de ecualización del histograma basado en mosaicos** es una técnica avanzada de procesamiento digital de imágenes que mejora el contraste de manera ***local y adaptativa***, dividiendo la imagen en regiones para un tratamiento individualizado.

---

## 📋 **Proceso de Implementación**

| **Paso** | **Proceso** | **Descripción** |
|:---:|:---|:---|
| <kbd>**1**</kbd> | **🔲 División de la imagen** | La imagen se particiona en múltiples mosaicos o tiles de tamaño uniforme |
| <kbd>**2**</kbd> | **📊 Cálculo individual** | Se computa la ecualización del histograma para cada mosaico de forma independiente |
| <kbd>**3**</kbd> | **🎯 Aplicación local** | Cada región recibe su propia transformación de intensidad optimizada |

---

## ⚖️ **Análisis Comparativo**

<table>
<tr>
<td width="50%">

### ✅ **VENTAJAS**

**1️⃣ Procesamiento localizado**  
`►` Calcula la ecualización de histograma en secciones distintas  
`►` Preserva características locales importantes

**2️⃣ Preservación de bordes**  
`►` Mantiene la definición de bordes en diferentes regiones  
`►` Evita la pérdida de detalles estructurales

**3️⃣ Mejora adaptativa**  
`►` Optimiza el contraste de manera local  
`►` Se adapta a las características de cada región

</td>
<td width="50%">

### ❌ **DESVENTAJAS**

**1️⃣ Amplificación de ruido**  
`►` Sobreamplifica el ruido en regiones homogéneas  
`►` Problema cuando el histograma está muy concentrado

**2️⃣ Artefactos visuales**  
`►` Diferencias de contraste entre mosaicos adyacentes  
`►` Puede crear efectos de "patchwork" no deseados

**3️⃣ Costo computacional**  
`►` Mayor tiempo de procesamiento  
`►` Requiere múltiples ecualizaciones independientes

</td>
</tr>
</table>

## 💡 **Conclusión**

<blockquote>

### **¿Cuándo es superior al método tradicional?**

El método de **`Tile-based Histogram Equalization`** supera al método tradicional cuando se procesan:

- 🌓 Imágenes con regiones de brillo **extremadamente variable**
- 🏙️ Escenas con áreas muy oscuras **Y** muy brillantes simultáneamente  
- 🔬 Imágenes complejas con **múltiples regiones de interés**

### **⚠️ Consideraciones importantes:**

- **Precaución:** En imágenes con grandes áreas uniformes, la amplificación del ruido puede degradar la calidad visual
- **Factor crítico:** La elección del tamaño de los mosaicos es crucial para balancear el mejoramiento local con la coherencia global

</blockquote>

````python
import math

def procesar_imagen_por_bloques(img_path: str, n_divisiones: int) -> Image.Image:
    """
    Divide una imagen en 'n' bloques, ecualiza el histograma de cada bloque
    individualmente y luego los une para formar una nueva imagen.

    Args:
        img_path (str): La ruta a la imagen de entrada.
        n_divisiones (int): El número de bloques en los que se dividirá la imagen.

    Returns:
        Image.Image: La imagen procesada con ecualización adaptativa.
    """
    try:
        img = Image.open(img_path).convert('RGB')
    except FileNotFoundError:
        print(f"Error: No se encontró el archivo en la ruta '{img_path}'")
        return None

    w, h = img.size

    # Si n <= 1, se realiza una ecualización global simple.
    if n_divisiones <= 1:
        return ImageOps.equalize(img)

    # Busca los factores de n_divisiones más cercanos a su raíz cuadrada
    # para que la cuadrícula sea lo más "cuadrada" posible.
    # Esto funciona para cualquier número, no solo para cuadrados perfectos (4, 9, 16...).
    cols = int(math.sqrt(n_divisiones))
    while n_divisiones % cols != 0:
        cols -= 1
    rows = n_divisiones // cols

    tile_w = w // cols
    tile_h = h // rows

    tiles = []
    # Itera sobre las filas y columnas para procesar cada bloque
    for r in range(rows):
        for c in range(cols):
            # Define la caja de recorte para el bloque actual
            box = (c * tile_w, r * tile_h, (c + 1) * tile_w, (r + 1) * tile_h)
            tile = img.crop(box)

            # Ecualiza el bloque y lo añade a la lista
            tile_eq = ImageOps.equalize(tile)
            tiles.append(tile_eq)

    # Crea una nueva imagen en blanco para pegar los bloques procesados
    new_img = Image.new('RGB', (w, h))

    # Itera sobre los bloques procesados y los pega en su posición correcta
    for i, tile in enumerate(tiles):
        r = i // cols
        c = i % cols
        new_img.paste(tile, (c * tile_w, r * tile_h))

    return new_img

# --- SCRIPT DE VISUALIZACIÓN ---

# Configuración
ruta_imagen = '2017_Brockton_Massachusetts'
divisiones_a_probar = [2, 4, 6, 9, 20]

# Crea una figura y un conjunto de subplots (ejes) en una sola fila
fig, axes = plt.subplots(1, len(divisiones_a_probar), figsize=(20, 5))

# Título principal para toda la figura
fig.suptitle('Comparación de Ecualización Adaptativa por Bloques', fontsize=18)

# Itera sobre la lista de divisiones y los ejes para procesar y mostrar cada imagen
for ax, n in zip(axes, divisiones_a_probar):
    imagen_procesada = procesar_imagen_por_bloques(ruta_imagen, n)

    if imagen_procesada:
        ax.imshow(imagen_procesada)
        ax.set_title(f'{n} Divisiones') # Añade el título a cada imagen
        ax.axis('off') # Oculta los ejes para un look más limpio

# Ajusta el espaciado para que no se solapen los títulos y muestra la figura
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
````

````python
def procesar_gris_por_bloques(img_path: str, n_divisiones: int) -> Image.Image:
    """
    Divide una imagen en escala de grises en 'n' bloques, ecualiza el
    histograma de cada bloque con cv2.equalizeHist y los une de nuevo.

    Args:
        img_path (str): La ruta a la imagen de entrada.
        n_divisiones (int): El número de bloques en los que se dividirá.

    Returns:
        Image.Image: La imagen procesada en escala de grises.
    """
    try:
        # Abrir la imagen directamente en escala de grises ('L' mode)
        img = Image.open(img_path).convert('L')
    except FileNotFoundError:
        print(f"Error: No se encontró el archivo en la ruta '{img_path}'")
        return None

    w, h = img.size

    if n_divisiones <= 1:
        # Ecualización global para el caso base
        img_np = np.array(img)
        img_eq_np = cv2.equalizeHist(img_np)
        return Image.fromarray(img_eq_np)

    # Calcular filas y columnas para que la cuadrícula sea lo más cuadrada posible
    cols = int(math.sqrt(n_divisiones))
    while n_divisiones % cols != 0:
        cols -= 1
    rows = n_divisiones // cols

    tile_w = w // cols
    tile_h = h // rows

    # Crea una imagen en blanco para pegar los resultados
    new_img = Image.new('L', (w, h))

    # Itera sobre las filas y columnas para procesar cada bloque
    for r in range(rows):
        for c in range(cols):
            # Recorta el bloque (tile)
            box = (c * tile_w, r * tile_h, (c + 1) * tile_w, (r + 1) * tile_h)
            tile = img.crop(box)

            # Procesa el bloque: PIL -> NumPy -> equalizeHist -> PIL
            tile_np = np.array(tile)
            tile_eq_np = cv2.equalizeHist(tile_np)
            tile_eq = Image.fromarray(tile_eq_np)

            # Pega el bloque procesado en la nueva imagen
            new_img.paste(tile_eq, box)

    return new_img

# --- SCRIPT DE VISUALIZACIÓN ---

# Configuración
ruta_imagen = '2017_Brockton_Massachusetts_gray.jpeg'
divisiones_a_probar = [2, 4, 6, 9, 20] # Valores que solicitaste

# Crea una figura y un conjunto de subplots en una fila horizontal
fig, axes = plt.subplots(1, len(divisiones_a_probar), figsize=(25, 5))

fig.suptitle('Comparación de Ecualización por Bloques (Escala de Grises)', fontsize=18)

# Itera para procesar y mostrar cada imagen
for ax, n in zip(axes, divisiones_a_probar):
    imagen_procesada = procesar_gris_por_bloques(ruta_imagen, n)

    if imagen_procesada:
        # Muestra la imagen. Es CRÍTICO usar cmap='gray' para verla correctamente.
        ax.imshow(imagen_procesada, cmap='gray')
        ax.set_title(f'{n} Divisiones')
        ax.axis('off')

# Ajusta el espaciado y muestra la figura final
plt.tight_layout(rect=[0, 0, 1, 0.94])
plt.show()
````

````python
import numpy as np
from PIL import Image
import cv2
import matplotlib.pyplot as plt

# ===================================================================
# TUS CLASES Y FUNCIONES (SE MANTIENEN EXACTAMENTE IGUAL)
# ===================================================================

class Tile:
    """Class to represent a tile of an image"""
    def __init__(self, image, position):
        self.image = image
        self.position = position

def split_image_to_tiles(img_path, n):
    """
    Split an image into n x n tiles
    """
    if isinstance(img_path, str):
        img = Image.open(img_path)
    else:
        img = img_path

    if img.mode != 'L':
        img = img.convert('L')

    width, height = img.size
    tile_width = width // n
    tile_height = height // n

    tiles = []

    for row in range(n):
        for col in range(n):
            left = col * tile_width
            top = row * tile_height
            right = left + tile_width
            bottom = top + tile_height

            if col == n - 1:
                right = width
            if row == n - 1:
                bottom = height

            tile = img.crop((left, top, right, bottom))
            tile_obj = Tile(tile, (row, col))
            tiles.append(tile_obj)

    return tiles

def join_tiles(tiles):
    """
    Join tiles back into a single image
    """
    if not tiles:
        return None

    n = int(np.sqrt(len(tiles)))
    sorted_tiles = sorted(tiles, key=lambda t: (t.position[0], t.position[1]))

    # Re-calculate final image size based on the actual tile sizes
    # This correctly handles images where dimensions aren't perfectly divisible
    grid_widths = [0] * n
    grid_heights = [0] * n
    for tile in sorted_tiles:
        row, col = tile.position
        w, h = tile.image.size
        grid_widths[col] = max(grid_widths[col], w)
        grid_heights[row] = max(grid_heights[row], h)

    max_width = sum(grid_widths)
    max_height = sum(grid_heights)

    new_img = Image.new('L', (max_width, max_height))

    y_offset = 0
    for row in range(n):
        x_offset = 0
        for col in range(n):
            # Find the tile for the current position
            tile_to_paste = next(t for t in sorted_tiles if t.position == (row, col))
            new_img.paste(tile_to_paste.image, (x_offset, y_offset))
            x_offset += grid_widths[col]
        y_offset += grid_heights[row]

    return new_img


def split_equalize_join_grey(img_path, n):
    """
    Split a grayscale image into tiles, apply histogram equalization to each tile,
    and join them back together
    """
    tiles = split_image_to_tiles(img_path, n)

    for tile in tiles:
        tile_array = np.array(tile.image)
        equalized_array = cv2.equalizeHist(tile_array)
        tile.image = Image.fromarray(equalized_array)

    result_image = join_tiles(tiles)
    return result_image

# ===================================================================
# NUEVO SCRIPT DE VISUALIZACIÓN HORIZONTAL
# ===================================================================

# --- Configuración ---
# ⚠️ Asegúrate de que la ruta a tu imagen sea la correcta
ruta_imagen = '2025_Tec_De_Monterrey_gray.jpg'
# Lista de divisiones por dimensión a probar
divisiones_a_probar = [2, 4, 6, 9, 12, 20]

# Carga la imagen original una sola vez
try:
    original_img = Image.open(ruta_imagen).convert('L')
except FileNotFoundError:
    print(f"Error: No se encontró el archivo en '{ruta_imagen}'")
    # Termina el script si no se encuentra la imagen
    exit()

# --- Creación de la figura ---
# El número de plots es la cantidad de pruebas + 1 para la original
num_plots = len(divisiones_a_probar) + 1
fig, axes = plt.subplots(1, num_plots, figsize=(28, 4)) # Figura ancha para que quepan todas

fig.suptitle('Comparación de Ecualización Adaptativa por Mosaicos', fontsize=18)

# --- Plot 1: Imagen Original ---
axes[0].imshow(original_img, cmap='gray')
axes[0].set_title('Original')
axes[0].axis('off')

# --- Plots Siguientes: Imágenes Procesadas ---
# Iteramos sobre el resto de los ejes y la lista de divisiones
for i, n in enumerate(divisiones_a_probar):
    ax = axes[i + 1] # Empezamos desde el segundo eje (índice 1)

    # Procesamos la imagen para el valor 'n' actual
    imagen_procesada = split_equalize_join_grey(original_img, n)

    if imagen_procesada:
        ax.imshow(imagen_procesada, cmap='gray')
        ax.set_title(f'{n}x{n} Mosaicos')
        ax.axis('off')

# --- Mostrar la figura ---
plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Ajusta el layout para dar espacio al supertítulo
plt.show()
````

````python
class Tile:
    """Class to represent a tile of an image"""
    def __init__(self, image, position):
        self.image = image
        self.position = position

def split_image_to_tiles(img_path, n):
    """
    Split an image into n x n tiles
    """
    # Load image
    if isinstance(img_path, str):
        img = Image.open(img_path)
    else:
        img = img_path

    # Convert to grayscale if needed
    if img.mode != 'L':
        img = img.convert('L')

    # Get image dimensions
    width, height = img.size

    # Calculate tile dimensions
    tile_width = width // n
    tile_height = height // n

    tiles = []

    for row in range(n):
        for col in range(n):
            # Calculate tile boundaries
            left = col * tile_width
            top = row * tile_height
            right = left + tile_width
            bottom = top + tile_height

            # Handle edge cases for last tiles
            if col == n - 1:
                right = width
            if row == n - 1:
                bottom = height

            # Crop the tile
            tile = img.crop((left, top, right, bottom))

            # Create Tile object with position info
            tile_obj = Tile(tile, (row, col))
            tiles.append(tile_obj)

    return tiles

def join_tiles(tiles):
    """
    Join tiles back into a single image
    """
    if not tiles:
        return None

    # Determine grid size (assuming square grid)
    n = int(np.sqrt(len(tiles)))

    # Sort tiles by position
    sorted_tiles = sorted(tiles, key=lambda t: (t.position[0], t.position[1]))

    # Get dimensions from first tile
    first_tile = sorted_tiles[0].image
    tile_width, tile_height = first_tile.size

    # Calculate dimensions for the final image
    max_width = 0
    max_height = 0

    for tile in sorted_tiles:
        row, col = tile.position
        w, h = tile.image.size
        if col == n - 1:  # Last column
            if max_width < (n - 1) * tile_width + w:
                max_width = (n - 1) * tile_width + w
        if row == n - 1:  # Last row
            if max_height < (n - 1) * tile_height + h:
                max_height = (n - 1) * tile_height + h

    # Create new image
    new_img = Image.new('L', (max_width, max_height))

    # Paste tiles
    for tile in sorted_tiles:
        row, col = tile.position
        x = col * tile_width
        y = row * tile_height
        new_img.paste(tile.image, (x, y))

    return new_img

def split_equalize_join_grey(img_path, n):
    """
    Split a grayscale image into tiles, apply histogram equalization to each tile,
    and join them back together
    """
    # Split image into tiles
    tiles = split_image_to_tiles(img_path, n)

    # Apply histogram equalization to each tile
    for tile in tiles:
        # Convert PIL Image to numpy array
        tile_array = np.array(tile.image)

        # Apply histogram equalization
        equalized_array = cv2.equalizeHist(tile_array)

        # Convert back to PIL Image
        tile.image = Image.fromarray(equalized_array)

    # Join tiles back together
    result_image = join_tiles(tiles)

    return result_image

# Aplicar ecualización adaptativa por mosaicos
n = 12  # Número de mosaicos por dimensión (12x12 = 144 mosaicos)
test_image = split_equalize_join_grey('2025_Tec_De_Monterrey_gray.jpg', n)

# Mostrar los resultados
plt.figure(figsize=(15, 7))

# Mostrar imagen original
plt.subplot(1, 2, 1)
original_img = Image.open('2025_Tec_De_Monterrey_gray.jpg')
plt.imshow(original_img, cmap='gray')
plt.title('Imagen Original en Escala de Grises')
plt.axis('off')

# Mostrar imagen ecualizada
plt.subplot(1, 2, 2)
plt.imshow(test_image, cmap='gray')
plt.title(f'Imagen Ecualizada ({n}x{n} mosaicos)')
plt.axis('off')

plt.tight_layout()
plt.show()
````

El **filtro de suavizado (smoothing)**(1) es una técnica usada para reducir las diferencias repentinas entre los bloques de una imagen (cambios de contraste, bordes duros o artefactos visibles). Aplica operaciones que disminuyen la variación de intensidad entre píxeles adyacentes, haciendo que sus transiciones sean más suaves.

Algunos de los filtros que podemos encontrar son:

    • filtro gaussiano, aplicaría un núcleo basado en una distribución normal a los valores de píxeles promedio y mantendría los detalles en las regiones de bajo gradiente.
    • filtro bilateral, que hace lo mismo que el anterior pero también incluye similitud de intensidad; permite preservar los bordes importantes mientras suaviza las áreas homogéneas.
    • filtro de mediana funciona bien al reemplazar cada píxel con la mediana de su vecindario, eliminando así los valores atípicos sin mucho desenfoque sobre todo en imágenes con mucho ruido.


El **blending(2)** es un método clave en la visión computacional para suavizar los cambios entre partes o secciones de una imagen, especialmente cuando hay diferencias de contraste, brillo o color. Su objetivo principal es obtener una apariencia visual suave eliminando los bordes repentinos o los artefactos entre las partes.

Algunas de las técnicas de blending son:

    • alpha blending - donde las intensidades de los píxeles en el área compartida entre las partes se mezclan de acuerdo con una cantidad relativa en el área de cambio.
    • pyramid blending - para mezclar partes en diferentes niveles de frecuencia, asegurándose de que los detalles finos y generales se junten de manera lógica. Esta técnica es particularmente útil en imágenes de alta resolución o cuando las diferencias entre las partes son grandes.

# SWAHE (Sliding Window Adaptive Histogram Equalization)

````python
def apply_swahe(image_array, window_size=20):
    """
    Aplica una implementación sencilla de SWAHE en una imagen en escala de grises.

    Args:
        image_array (np.ndarray): La imagen de entrada como un array de NumPy.
        window_size (int): El tamaño de la ventana deslizante (debe ser un número par).

    Returns:
        np.ndarray: La imagen con el contraste mejorado.
    """
    if window_size % 2 != 0:
        raise ValueError("El tamaño de la ventana debe ser un número par.")

    # Crear una matriz para la imagen de salida
    output_image = np.zeros_like(image_array, dtype=np.uint8)

    h, w = image_array.shape
    pad_size = window_size // 2

    # Rellenar la imagen con ceros para manejar los bordes
    padded_image = np.pad(image_array, pad_size, mode='edge')

    # Recorrer cada píxel de la imagen original
    for i in range(h):
        for j in range(w):
            # 1. Definir la ventana local (vecindad)
            window = padded_image[i:i + window_size, j:j + window_size]

            # 2. Calcular el histograma y el CDF para la ventana
            hist, _ = np.histogram(window.flatten(), 256, [0, 256])
            cdf = hist.cumsum()
            cdf_normalized = cdf * hist.max() / cdf.max()

            # 3. Aplicar la ecualización al píxel central
            pixel_value = image_array[i, j]
            equalized_value = cdf_normalized[pixel_value]
            output_image[i, j] = equalized_value

    return output_image

````

# CLAHE (Contrast Limited Adaptive Histogram Equalization) - Conceptualización

CLAHE is an advanced form of Adaptive Histogram Equalization (AHE) that overcomes the limitation of noise amplification in AHE. While AHE calculates and applies histogram equalization to small regions (tiles) of the image, it can over-amplify contrast in uniform areas, making noise more prominent.

CLAHE addresses this by introducing a **contrast limiting step**. Before computing the Cumulative Distribution Function (CDF) for each tile, the histogram of that tile is "clipped" at a predefined limit. Any histogram bins exceeding this limit are redistributed among the other bins. This process prevents excessive spikes in the histogram, which in turn limits the steepness of the CDF and thus limits the contrast stretching.

After the contrast-limited histogram equalization is applied to each tile, a bilinear interpolation is typically used to smooth out the transitions between adjacent tiles, reducing the appearance of artificial boundaries or checkerboard effects that can occur with simple tile-based methods.

**Key differences from basic Histogram Equalization and AHE:**

*   **Basic HE:** Applies a single histogram equalization to the entire image, often failing to enhance local contrast in areas that are significantly darker or brighter than the overall image.
*   **AHE:** Applies histogram equalization to local regions, improving local contrast but prone to noise amplification.
*   **CLAHE:** Applies contrast-limited histogram equalization to local regions and uses interpolation to smooth boundaries, providing good local contrast enhancement while limiting noise amplification and reducing tile artifacts.

CLAHE is particularly useful for images with varying lighting conditions or medical images where enhancing fine details in specific regions is critical.

````python
# Function to display images
def display_images(images, titles):
    """Displays a list of images with titles."""
    num_images = len(images)
    plt.figure(figsize=(5 * num_images, 5))
    for i in range(num_images):
        plt.subplot(1, num_images, i + 1)
        # Check if the image is a PIL Image or a NumPy array
        if isinstance(images[i], Image.Image):
             # Use image mode for PIL Image
             if images[i].mode == 'L':
                  plt.imshow(images[i], cmap='gray')
             else:
                  plt.imshow(images[i])
        elif isinstance(images[i], np.ndarray):
             # For NumPy arrays, check dimensions to infer color or grayscale
             if images[i].ndim == 2:
                  plt.imshow(images[i], cmap='gray')
             else: # Assuming 3 dimensions for color (H, W, C)
                  # OpenCV loads images as BGR, matplotlib expects RGB
                  plt.imshow(cv2.cvtColor(images[i], cv2.COLOR_BGR2RGB))
        else:
             print(f"Warning: Unrecognized image format for image {i}")
             plt.imshow(images[i]) # Attempt to display anyway

        plt.title(titles[i])
        plt.axis('off')
    plt.tight_layout()
    plt.show()

````

````python
# --- Load a sample image (using the downloaded Brockton image) ---
image_path = '2017_Brockton_Massachusetts' # Path to your image file
output_gray_path = '2017_Brockton_Massachusetts_gray_for_clahe.jpg'

# Ensure a grayscale version exists for processing
if not os.path.exists(output_gray_path):
    print(f"'{output_gray_path}' not found. Creating a grayscale version...")
    try:
        original_image_color = Image.open(image_path)
        original_image_gray_pil = original_image_color.convert('L')
        original_image_gray_pil.save(output_gray_path)
        print(f"'{output_gray_path}' created.")
    except FileNotFoundError:
        print(f"Error: Original image '{image_path}' not found.")
        # Exit if the original image is not found
        exit()

try:
    # Load the grayscale image using OpenCV
    img_gray = cv2.imread(output_gray_path, cv2.IMREAD_GRAYSCALE)

    # --- 1. Basic Histogram Equalization ---
    img_basic_eq = cv2.equalizeHist(img_gray)

    # --- 2. CLAHE Implementation using OpenCV ---
    # Create a CLAHE object
    # clipLimit: Threshold for contrast limiting (e.g., 2.0, 3.0, 4.0)
    # tileGridSize: Size of the grid for histogram equalization (e.g., (8, 8), (16, 16))
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))

    # Apply CLAHE
    img_clahe = clahe.apply(img_gray)

    # --- Display Results ---
    display_images([img_gray, img_basic_eq, img_clahe],
                   ['Original Grayscale', 'Basic HE', 'CLAHE'])

except Exception as e:
    print(f"An error occurred during image processing: {e}")
````
