````python
from PIL import Image
from PIL.ExifTags import TAGS
from tabulate import tabulate
import textwrap

def display_exif_tabulate(image_path):
    img_pil = Image.open(image_path)
    exif_data = img_pil._getexif()

    if not exif_data:
        print("No se encontraron datos EXIF")
        return

    # Preparar datos para la tabla
    table_data = []
    for tag_id, value in exif_data.items():
        tag = TAGS.get(tag_id, tag_id)

        # Limitar longitud de valores muy largos
        value_str = str(value)
        if len(value_str) > 50:
            value_str = textwrap.fill(value_str, width=50)

        # Agregar emojis para ciertas categorías
        emoji = ""
        if 'Date' in tag:
            emoji = "📅"
        elif 'GPS' in tag:
            emoji = "📍"
        elif tag in ['Make', 'Model']:
            emoji = "📷"
        elif tag in ['ISO', 'ExposureTime', 'FNumber']:
            emoji = "⚙️"

        table_data.append([f"{emoji} {tag}", value_str])

    # Mostrar tabla
    print("\n" + "="*60)
    print("📸 DATOS EXIF DE LA IMAGEN 📸".center(60))
    print("="*60 + "\n")

    print(tabulate(table_data,
                   headers=['Propiedad', 'Valor'],
                   tablefmt='fancy_grid',
                   colalign=('left', 'left')))

# Usar:
display_exif_tabulate('2017_Brockton_Massachusetts')
````

## Histogram Equalization<a class="anchor" id="hist" ></a>

Let's look at the distribution of pixel values in the image.

Ideally, the image should have relatively even distribution of values, indicating good contrast and making it easier to extract analytical information.

An easy way to check this is to plot a histogram.

````python
# Plot a histogram - we need to use ravel to "flatten" the 3 dimensions
plt.hist(original_image.ravel())
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image

def plot_minimalist_histogram(image_path):
    # Configuración minimalista
    plt.rcParams['figure.facecolor'] = '#f8f9fa'
    plt.rcParams['axes.facecolor'] = '#ffffff'
    plt.rcParams['font.family'] = 'sans-serif'
    plt.rcParams['font.sans-serif'] = ['Arial', 'DejaVu Sans']

    # Cargar imagen
    img = Image.open(image_path)
    img_array = np.array(img)

    # Crear figura
    fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    fig.patch.set_facecolor('#f8f9fa')

    # Configuración de colores suaves
    colors = {
        'R': {'main': '#ff6b6b', 'light': '#ffe0e0'},
        'G': {'main': '#51cf66', 'light': '#d0f4d0'},
        'B': {'main': '#339af0', 'light': '#d0e7ff'}
    }

    titles = ['Canal Rojo', 'Canal Verde', 'Canal Azul']

    for idx, (ax, channel_name, title) in enumerate(zip(axes, ['R', 'G', 'B'], titles)):
        # Obtener datos del canal
        channel_data = img_array[:,:,idx].ravel()

        # Crear histograma con estilo elegante
        n, bins, patches = ax.hist(channel_data, bins=256,
                                   color=colors[channel_name]['light'],
                                   edgecolor=colors[channel_name]['main'],
                                   linewidth=0.5, alpha=0.7)

        # Agregar línea suave de tendencia
        from scipy.ndimage import gaussian_filter1d
        smoothed = gaussian_filter1d(n, sigma=2)
        bin_centers = (bins[:-1] + bins[1:]) / 2
        ax.plot(bin_centers, smoothed, color=colors[channel_name]['main'],
               linewidth=2, alpha=0.9)

        # Estilizar ejes
        ax.set_title(title, fontsize=12, fontweight='600', pad=15)
        ax.set_xlabel('Intensidad', fontsize=10)
        ax.set_ylabel('Frecuencia', fontsize=10)

        # Remover bordes superiores y derechos
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
        ax.spines['left'].set_linewidth(0.5)
        ax.spines['bottom'].set_linewidth(0.5)

        # Grid sutil
        ax.grid(True, alpha=0.1, linestyle='-', linewidth=0.5)
        ax.set_axisbelow(True)

        # Agregar estadística clave de forma sutil
        mean_val = np.mean(channel_data)
        ax.text(0.98, 0.98, f'μ = {mean_val:.1f}',
               transform=ax.transAxes,
               fontsize=10, ha='right', va='top',
               bbox=dict(boxstyle='round,pad=0.3',
                        facecolor='white',
                        edgecolor=colors[channel_name]['main'],
                        alpha=0.8))

    # Título general
    fig.suptitle(f'Análisis de Histograma: {image_path}',
                fontsize=14, fontweight='bold', y=1.05)

    plt.tight_layout()
    plt.show()

# Usar
plot_minimalist_histogram('2017_Brockton_Massachusetts')
````

Another useful way to visualize the statistics of an image is as a cumulative distribution function (CDF) plot. Which shows the cumulative pixel intensity frequencies from 0 to 255.

````python
plt.hist(original_image.ravel(), bins=255, cumulative=True)
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
from scipy.ndimage import gaussian_filter1d

def plot_cumulative_histogram_beautiful(image_path):
    # Configuración minimalista
    plt.rcParams['figure.facecolor'] = '#f8f9fa'
    plt.rcParams['axes.facecolor'] = '#ffffff'
    plt.rcParams['font.family'] = 'sans-serif'
    plt.rcParams['font.sans-serif'] = ['Arial', 'DejaVu Sans']

    # Cargar imagen
    img = Image.open(image_path)
    img_array = np.array(img)

    # Crear figura con subplots
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    fig.patch.set_facecolor('#f8f9fa')

    # Configuración de colores suaves
    colors = {
        'R': {'main': '#ff6b6b', 'light': '#ffe0e0', 'dark': '#e55555'},
        'G': {'main': '#51cf66', 'light': '#d0f4d0', 'dark': '#40b855'},
        'B': {'main': '#339af0', 'light': '#d0e7ff', 'dark': '#228ae6'},
        'All': {'main': '#868e96', 'light': '#e9ecef', 'dark': '#495057'}
    }

    # ============= SUBPLOT 1: Histograma Acumulativo Combinado =============
    ax1 = axes[0, 0]

    # Datos para todos los canales
    for idx, (channel_name, title) in enumerate(zip(['R', 'G', 'B'],
                                                     ['Rojo', 'Verde', 'Azul'])):
        channel_data = img_array[:,:,idx].ravel()

        # Crear histograma acumulativo
        counts, bins, patches = ax1.hist(channel_data, bins=256,
                                         cumulative=True,
                                         density=True,
                                         histtype='step',
                                         color=colors[channel_name]['main'],
                                         linewidth=2.5,
                                         alpha=0.8,
                                         label=f'Canal {title}')

    ax1.set_title('Distribución Acumulativa RGB', fontsize=13, fontweight='600', pad=15)
    ax1.set_xlabel('Intensidad de Píxel', fontsize=11)
    ax1.set_ylabel('Probabilidad Acumulada', fontsize=11)
    ax1.legend(loc='lower right', frameon=True, fancybox=True, shadow=True)
    ax1.grid(True, alpha=0.15, linestyle='-', linewidth=0.5)
    ax1.set_xlim([0, 255])
    ax1.set_ylim([0, 1.05])

    # Agregar líneas de referencia
    ax1.axhline(y=0.5, color='gray', linestyle='--', alpha=0.3, linewidth=1)
    ax1.axhline(y=0.25, color='gray', linestyle=':', alpha=0.2, linewidth=1)
    ax1.axhline(y=0.75, color='gray', linestyle=':', alpha=0.2, linewidth=1)

    # ============= SUBPLOT 2: Histograma Acumulativo Total (Luminosidad) =============
    ax2 = axes[0, 1]

    # Calcular luminosidad
    luminosity = np.mean(img_array, axis=2).ravel()

    # Histograma acumulativo con relleno
    counts, bins, patches = ax2.hist(luminosity, bins=256,
                                     cumulative=True,
                                     density=True,
                                     color=colors['All']['light'],
                                     edgecolor=colors['All']['dark'],
                                     linewidth=1.5,
                                     alpha=0.7)

    # Línea suavizada
    bin_centers = (bins[:-1] + bins[1:]) / 2
    smoothed = gaussian_filter1d(counts, sigma=1)
    ax2.plot(bin_centers, smoothed, color=colors['All']['dark'],
            linewidth=3, alpha=0.9, label='Curva suavizada')

    # Encontrar percentiles
    percentiles = [25, 50, 75]
    for p in percentiles:
        val = np.percentile(luminosity, p)
        ax2.axvline(x=val, color='red', linestyle='--', alpha=0.5, linewidth=1)
        ax2.text(val, 0.05, f'P{p}', rotation=90, fontsize=9,
                ha='right', va='bottom', color='red', alpha=0.7)

    ax2.set_title('Distribución Acumulativa de Luminosidad', fontsize=13, fontweight='600', pad=15)
    ax2.set_xlabel('Intensidad de Píxel', fontsize=11)
    ax2.set_ylabel('Probabilidad Acumulada', fontsize=11)
    ax2.legend(loc='lower right', frameon=True, fancybox=True, shadow=True)
    ax2.grid(True, alpha=0.15, linestyle='-', linewidth=0.5)
    ax2.set_xlim([0, 255])
    ax2.set_ylim([0, 1.05])

    # ============= SUBPLOTS 3-5: Canales Individuales con Área =============
    channels_data = [
        (img_array[:,:,0].ravel(), 'R', 'Canal Rojo'),
        (img_array[:,:,1].ravel(), 'G', 'Canal Verde'),
        (img_array[:,:,2].ravel(), 'B', 'Canal Azul')
    ]

    for idx, (data, color_key, title) in enumerate(channels_data):
        if idx < 2:
            ax = axes[1, idx]
        else:
            # Usar el último subplot disponible para el canal azul
            ax = plt.subplot2grid((2, 3), (1, 2), colspan=1, fig=fig)

        # Calcular histograma acumulativo
        counts, bins = np.histogram(data, bins=256)
        cumulative = np.cumsum(counts)
        cumulative = cumulative / cumulative[-1]  # Normalizar

        # Plot con relleno gradiente
        bin_centers = (bins[:-1] + bins[1:]) / 2
        ax.fill_between(bin_centers, 0, cumulative,
                       color=colors[color_key]['light'],
                       alpha=0.5)
        ax.plot(bin_centers, cumulative,
               color=colors[color_key]['main'],
               linewidth=2.5, alpha=0.9)

        # Agregar puntos en cuartiles
        quartiles = [25, 50, 75]
        for q in quartiles:
            val = np.percentile(data, q)
            y_val = np.interp(val, bin_centers, cumulative)
            ax.scatter(val, y_val, color=colors[color_key]['dark'],
                      s=50, zorder=5, alpha=0.8)
            ax.annotate(f'{q}%', xy=(val, y_val),
                       xytext=(5, 5), textcoords='offset points',
                       fontsize=9, color=colors[color_key]['dark'],
                       fontweight='bold')

        # Estadísticas
        mean_val = np.mean(data)
        median_val = np.median(data)
        std_val = np.std(data)

        stats_text = f'μ={mean_val:.1f}\nσ={std_val:.1f}'
        ax.text(0.98, 0.05, stats_text,
               transform=ax.transAxes,
               fontsize=10, ha='right', va='bottom',
               bbox=dict(boxstyle='round,pad=0.5',
                        facecolor='white',
                        edgecolor=colors[color_key]['main'],
                        alpha=0.9))

        ax.set_title(title, fontsize=12, fontweight='600', pad=10)
        ax.set_xlabel('Intensidad', fontsize=10)
        ax.set_ylabel('Probabilidad Acumulada', fontsize=10)
        ax.set_xlim([0, 255])
        ax.set_ylim([0, 1.05])

        # Estilo minimalista
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
        ax.spines['left'].set_linewidth(0.5)
        ax.spines['bottom'].set_linewidth(0.5)
        ax.grid(True, alpha=0.1, linestyle='-', linewidth=0.5)
        ax.set_axisbelow(True)

    # Título general
    fig.suptitle(f'📊 Análisis de Distribución Acumulativa: {image_path}',
                fontsize=15, fontweight='bold', y=1.02)

    plt.tight_layout()
    plt.show()

# Usar
plot_cumulative_histogram_beautiful('2017_Brockton_Massachusetts')
````

The histogram and CDF for our image show pretty uneven distribution - there's a lot of contrast in the image. Ideally we should equalize the values in the images we want to analyse to try to make our images more consistent in terms of the shapes they contain irrespective of light levels.

Histogram equalization is often used to improve the statistics of images. In simple terms, the histogram equalization algorithm attempts to adjust the pixel values in the image to create a more uniform distribution. The code in the cell below uses the  **exposure.equalize_hist** method from the **skimage** package to equalize the image.  

````python
original_image_eq = exposure.equalize_hist(original_image)

fig = plt.figure(figsize=(16, 8))

# Subplot for original image
a=fig.add_subplot(1,2,1)
imgplot = plt.imshow(original_image)
a.set_title('Before')

# Subplot for processed image
a=fig.add_subplot(1,2,2)
imgplot = plt.imshow(original_image_eq)
a.set_title('After')

plt.show()
````

As with most image operations, there's more than one way to do this. For example, you could also use the **PIL.ImgOps.equalize** method:

````python

# Equalize the image - but we need to convert the numpy array back to the PIL image format
image_PIL_eq = ImageOps.equalize(Image.fromarray(original_image))

# Display using matplotlib

# Create a figure
fig = plt.figure(figsize=(16, 8))

# Subplot for original image
a=fig.add_subplot(1,2,1)
imgplot = plt.imshow(original_image)
a.set_title('Before')

# Subplot for processed image
a=fig.add_subplot(1,2,2)
imgplot = plt.imshow(image_PIL_eq)
a.set_title('After')

plt.show()
````

````python
# Display histograms
img_eq = np.array(image_PIL_eq)

# Create a figure
fig = plt.figure(figsize=(16, 8))

# Subplot for original image
a=fig.add_subplot(1,2,1)
imgplot = plt.hist(img_eq.ravel())
a.set_title('Histogram')

# Subplot for processed image
a=fig.add_subplot(1,2,2)
imgplot = plt.hist(img_eq.ravel(), bins=255, cumulative=True)
a.set_title('CDF')

plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
from scipy.ndimage import gaussian_filter1d

# 2. FUNCIÓN PARA CREAR LOS PLOTS
def plot_histogram_vs_cdf_beautiful(image_PIL_eq):
    """
    Crea visualización hermosa de Histograma y CDF
    """
    # Configuración minimalista
    plt.rcParams['figure.facecolor'] = '#f8f9fa'
    plt.rcParams['axes.facecolor'] = '#ffffff'
    plt.rcParams['font.family'] = 'sans-serif'

    # Convertir imagen a array
    img_eq = np.array(image_PIL_eq)
    data = img_eq.ravel()

    # Crear figura
    fig = plt.figure(figsize=(16, 8))
    fig.patch.set_facecolor('#f8f9fa')

    # Colores elegantes
    color_hist = '#3498db'  # Azul suave
    color_cdf = '#e74c3c'   # Rojo suave

    # ================== SUBPLOT 1: HISTOGRAMA ==================
    ax1 = fig.add_subplot(1, 2, 1)

    # Histograma con estilo
    n, bins, patches = ax1.hist(data, bins=256,
                                color='#bdd4e7',
                                edgecolor='#3498db',
                                linewidth=0.5,
                                alpha=0.8)

    # Aplicar gradiente de color a las barras
    for i, patch in enumerate(patches):
        intensity = i / len(patches)
        patch.set_facecolor(plt.cm.Blues(0.3 + intensity * 0.5))

    # Línea suavizada
    smoothed = gaussian_filter1d(n, sigma=2)
    bin_centers = (bins[:-1] + bins[1:]) / 2
    ax1.plot(bin_centers, smoothed, color=color_hist,
            linewidth=2.5, alpha=0.9, label='Tendencia suavizada')

    # Estadísticas
    mean_val = np.mean(data)
    median_val = np.median(data)
    std_val = np.std(data)

    # Líneas de estadísticas
    ax1.axvline(mean_val, color='#2ecc71', linestyle='--',
               linewidth=2, alpha=0.7, label=f'Media: {mean_val:.1f}')
    ax1.axvline(median_val, color='#f39c12', linestyle='--',
               linewidth=2, alpha=0.7, label=f'Mediana: {median_val:.1f}')

    # Configuración del subplot
    ax1.set_title('📊 Histograma', fontsize=14, fontweight='600', pad=15)
    ax1.set_xlabel('Intensidad de Píxel', fontsize=11)
    ax1.set_ylabel('Frecuencia', fontsize=11)
    ax1.legend(loc='upper right')
    ax1.grid(True, alpha=0.15)
    ax1.set_xlim([0, 255])

    # Remover bordes superiores
    ax1.spines['top'].set_visible(False)
    ax1.spines['right'].set_visible(False)

    # ================== SUBPLOT 2: CDF ==================
    ax2 = fig.add_subplot(1, 2, 2)

    # Calcular y plotear CDF
    n_cdf, bins_cdf, patches_cdf = ax2.hist(data, bins=255,
                                            cumulative=True,
                                            density=True,
                                            histtype='step',
                                            color=color_cdf,
                                            linewidth=2.5,
                                            alpha=0.9)

    # Rellenar área bajo la curva
    ax2.fill_between(bins_cdf[:-1], 0, n_cdf,
                    alpha=0.2, color=color_cdf)

    # Marcar percentiles
    percentiles = [25, 50, 75]
    colors_p = ['#9b59b6', '#2ecc71', '#f39c12']

    for p, color_p in zip(percentiles, colors_p):
        val = np.percentile(data, p)
        y_val = p / 100

        ax2.axhline(y=y_val, color=color_p, linestyle=':',
                   alpha=0.3, linewidth=1)
        ax2.axvline(x=val, color=color_p, linestyle=':',
                   alpha=0.3, linewidth=1)

        ax2.scatter(val, y_val, color=color_p, s=60,
                   zorder=5, alpha=0.8, edgecolor='white', linewidth=2)

        ax2.annotate(f'P{p}', xy=(val, y_val),
                    xytext=(8, 8), textcoords='offset points',
                    fontsize=9, color=color_p, fontweight='bold')

    # Configuración del subplot
    ax2.set_title('📈 CDF (Distribución Acumulativa)',
                 fontsize=14, fontweight='600', pad=15)
    ax2.set_xlabel('Intensidad de Píxel', fontsize=11)
    ax2.set_ylabel('Probabilidad Acumulada', fontsize=11)
    ax2.set_xlim([0, 255])
    ax2.set_ylim([0, 1.05])
    ax2.grid(True, alpha=0.15)

    # Remover bordes superiores
    ax2.spines['top'].set_visible(False)
    ax2.spines['right'].set_visible(False)

    # Título general
    fig.suptitle('🎨 Análisis de Distribución - Imagen Ecualizada',
                fontsize=16, fontweight='bold', y=0.98)

    plt.tight_layout()
    plt.show()  # ¡IMPORTANTE! Esto muestra el plot

# 3. EJECUTAR LA FUNCIÓN
plot_histogram_vs_cdf_beautiful(image_PIL_eq)
````

The pixel intensities are more evenly distributed in the equalized image. In particular, the cumulative density shows a straight diagonal cumulation; which is a good sign that the pixel intensity values have been equalized.

````python
img_n = skimage.util.random_noise(original_image)
plt.imshow(img_n)
````

### Using a Gaussian Filter
A Gaussian filter applies a weighted average (mean) value for pixels based on the pixels that surround them.

````python
img3_gauss = gauss(img_n, sigma=1)
plt.imshow(img3_gauss)
````

### Using a Median Filter
The Gaussian filter results in a blurred image, which may actually be better for feature extraction as it makes it easier to find contrasting areas. If it's too blurred, we could try a median filter, which as the name suggests applies the median value to pixels based on the pixels around them.

````python
img_med = med(img_n, size=2)
plt.imshow(img_med)
````
